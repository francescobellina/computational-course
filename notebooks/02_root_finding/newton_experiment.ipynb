# Metodo di Newton: esperimento numerico

## 1. Obiettivo

Studiare il metodo di Newton su $f(x)=\cos(x)-x$, verificare numericamente la convergenza
e visualizzare l'andamento del residuo.


## 2. Teoria

L'iterazione di Newton è

$$
x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}.
$$

Per il problema scelto:

$$
f(x)=\cos(x)-x, \qquad f'(x)=-\sin(x)-1.
$$


## 3. Implementazione

L'algoritmo viene importato dal package del progetto, non riscritto nel notebook.


```python
import numpy as np
import matplotlib.pyplot as plt

from computational_course.root_finding import newton
```

## 4. Esperimento

Usiamo $x_0=1$ e tolleranza $10^{-12}$.


```python
f = lambda x: np.cos(x) - x
df = lambda x: -np.sin(x) - 1.0

root, history = newton(f, df, 1.0, tol=1e-12)
residuals = np.array([abs(f(x)) for x in history])

root, len(history) - 1
```

## 5. Risultati


```python
print(f"Radice approssimata: {root:.12f}")
print(f"Residuo finale:      {abs(f(root)):.3e}")
print(f"Iterazioni:          {len(history) - 1}")
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.semilogy(range(len(residuals)), residuals, marker="o")
ax.set_xlabel("Iterazione n")
ax.set_ylabel(r"$|f(x_n)|$")
ax.set_title("Convergenza del metodo di Newton")
ax.grid(True, which="both", alpha=0.3)
plt.show()
```

## 6. Discussione

Il residuo diminuisce molto rapidamente dopo le prime iterazioni. Questo è coerente con la
convergenza quadratica locale attesa per una radice semplice quando il punto iniziale è
sufficientemente vicino alla soluzione.
